```python
obj = DataIngestionConfig()
```

```python
import pandas as pd
```

```python
ident_df = pd.read_csv(obj.train_identity_file_path)
trans_df = pd.read_csv(obj.train_transaction_file_path)
```

```python
merged_df = pd.merge(trans_df, ident_df, on="TransactionID", how="left")
```

```python
df = merged_df
```

```python
df["DT_hours"] = df["TransactionDT"] / 3600
df["DT_days"] = df["TransactionDT"] / 86400
df["DT_weeks"] = df["TransactionDT"] / (86400 * 7)
```

```python
df["hour"] = (df["TransactionDT"] // 3600) % 24
```

```python
df["day_of_week"] = (df["TransactionDT"] // 86400) % 7
```

```python
df["day"] = df["TransactionDT"] // 86400
```

```python
df["time_bucket"] = pd.cut(
    df["hour"],
    bins=[0, 6, 12, 18, 24],
    labels=["night", "morning", "afternoon", "evening"],
)
```

```python
fraud_df = df[df["isFraud"] == 1].reset_index(drop=True)
```
